```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns 

# load the dataset
df = pd.read_csv(r"C:\Users\ASUS\Projects\Customer-Data-Analysis-and-Churn-Prediction\data\Telco-Customer-Churn.csv")

df.info()
```

```python
# display how many missing values exist in each column
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
# calculate the percentage of missing values
missing_percentage = (df.isnull().sum() / len(df)) * 100

print(missing_percentage[missing_percentage > 0])

# Series (Pandas' one dimensional data structure)
# [] series is empty, no columns with missing values
# dtype: float64 - Series is designed to hold floating point for percentages
```

```python
df['TotalCharges'] = pd.to_numeric(
    df['TotalCharges'],
    errors='coerce'
)

# errors='coerce' - converts values that cant be interpreted as numbers into missing values
```

```python
df['TotalCharges'].isnull().sum()

# output means 11 missing values in the dataset
```
